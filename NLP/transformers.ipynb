Tokenizer Live Demo

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
sequence = "Using a Transformer network is simple"

tokens = tokenizer.tokenize(sequence)
print("tokens:", tokens)

ids = tokenizer.convert_tokens_to_ids(tokens)
print("input IDs:", ids)

inputs = tokenizer(sequence)
print("full encoding:", inputs)

decoded = tokenizer.decode(inputs["input_ids"])
print("decoded:", decoded)
```

Tokenizer Comparison: BERT vs GPT-2

```python
bert_tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
gpt2_tokenizer = AutoTokenizer.from_pretrained("gpt2")

sentence = "Transformers is amazing"

bert_tokens = bert_tokenizer.tokenize(sentence)
gpt2_tokens = gpt2_tokenizer.tokenize(sentence)

print("BERT  tokens:", bert_tokens)
print("BERT  ids   :", bert_tokenizer.convert_tokens_to_ids(bert_tokens))
print("GPT-2 tokens:", gpt2_tokens)
print("GPT-2 ids   :", gpt2_tokenizer.convert_tokens_to_ids(gpt2_tokens))
```

```python
from sentence_transformers import SentenceTransformer, util

model = SentenceTransformer("all-MiniLM-L6-v2")

# Wrap your keywords in meaningful structural contexts
phrases = {
    "messi": "Lionel Messi is a famous professional football player.",
    "football": "Football is a popular team sport played with a ball.",
    "Hockey": "Hockey is a team sport played on ice with sticks."
}

vecs = {k: model.encode(v, convert_to_tensor=True) for k, v in phrases.items()}
print(vecs)
# Compute cosine similarities
sim_football = util.cos_sim(vecs["messi"], vecs["football"])
sim_hockey = util.cos_sim(vecs["messi"], vecs["Hockey"])

print(f"messi vs football: {sim_football.item():.4f}")  # Significantly higher
print(f"messi vs Hockey  : {sim_hockey.item():.4f}")    # Significantly lower
```

```python
word1 = "king"
word2 = "man"
word3 = "queen"
word4 = "woman"

word5 = "football"

# king - man + woman = queen

embd1 = model.encode(word1)
embd2 = model.encode(word2)
embd3 = model.encode(word3)
embd4 = model.encode(word4)
embd5 = model.encode(word5)
```

```python
result = embd1 - embd2 + embd4
```

```python
sim1 = util.cos_sim(embd3, result)
sim2 = util.cos_sim(embd5, result)

print(sim1)
print(sim2)
```
