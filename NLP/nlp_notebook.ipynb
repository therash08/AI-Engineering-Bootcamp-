```python
"""
Shows word tokenization, sentence tokenization, and the difference
between a naive split() and NLTK's smart word_tokenize().

NLTK downloads required (auto-downloaded on first run):
    punkt_tab, punkt
"""

import nltk

# Auto-download required NLTK data (safe to run multiple times)
print("Downloading NLTK data packages...")
nltk.download("punkt",       quiet=True)
nltk.download("punkt_tab",   quiet=True)
print("✅ NLTK data ready\n")

from nltk.tokenize import (
    word_tokenize,       # Smart word tokenizer (handles punctuation)
    sent_tokenize,       # Sentence tokenizer
    TweetTokenizer,      # Good for social media text
    MWETokenizer,        # Multi-word expressions ("New York" as one token)
)
```

```python
# ── SECTION 1: Word Tokenization ─────────────────────────────────
print("=" * 60)
print("SECTION 1: Word Tokenization")
print("=" * 60)

sample_text = "The quick brown fox doesn't jump over the lazy dog! It ran away at 3.5 mph..."

print(f"\nOriginal text:\n  '{sample_text}'\n")

# Method 1: Naive Python split — WRONG way
naive_tokens = sample_text.split()
print(f"1. Naive split()  → {len(naive_tokens)} tokens")
print(f"   {naive_tokens}\n")
print("   ⚠️  Problem: 'dog!' mixed with punctuation\n")

# Method 2: NLTK word_tokenize — CORRECT way
nltk_tokens = word_tokenize(sample_text)
print(f"2. word_tokenize()  → {len(nltk_tokens)} tokens")
print(f"   {nltk_tokens}\n")
print("   ✅ Punctuation separated. \"doesn't\" → [\"does\", \"n't\"] (smart!)\n")
```

```python
# ── SECTION 2: Sentence Tokenization ─────────────────────────────
print("=" * 60)
print("SECTION 2: Sentence Tokenization")
print("=" * 60)

multi_sent = (
    "Dr. Smith works at N.Y.U. He published a paper in 2024. "
    "The paper was titled 'NLP is amazing!' and covered tokenization. "
    "Mr. Brown reviewed it; he gave it 9.5/10."
)

print(f"\nMulti-sentence text:\n  '{multi_sent}'\n")

sentences = sent_tokenize(multi_sent)
print(f"sent_tokenize() found {len(sentences)} sentences:\n")
for i, sent in enumerate(sentences, 1):
    print(f"  Sentence {i}: {sent}")

print("\n  ✅ 'Dr.' and 'N.Y.U.' NOT treated as sentence ends — smart!")
print("  ✅ '9.5/10' NOT treated as sentence end")
```

```python
# ── SECTION 3: Tweet Tokenizer ────────────────────────────────────
print("\n" + "=" * 60)
print("SECTION 3: Tweet Tokenizer (social media text)")
print("=" * 60)

tweet = "OMG I LOVE this!! 😍 #NLP #Python @BrainStation23 check this url: https://t.co/xyz"
print(f"\nTweet text:\n  '{tweet}'\n")

tweet_tokenizer = TweetTokenizer(preserve_case=False, strip_handles=True, reduce_len=True)
tweet_tokens = tweet_tokenizer.tokenize(tweet)
print(f"TweetTokenizer → {len(tweet_tokens)} tokens:")
print(f"  {tweet_tokens}")
print("\n  ✅ Handles hashtags, mentions, emojis, URLs specially")
```

```python
# ── SECTION 4: Comparison Summary ────────────────────────────────
print("\n" + "=" * 60)
print("SECTION 4: Token Count Comparison")
print("=" * 60)

test = "Running foxes don't eat quickly. They're beautiful!"
print(f"\nTest text: '{test}'")
print(f"\n  split()          : {len(test.split())} tokens → {test.split()}")
print(f"  word_tokenize()  : {len(word_tokenize(test))} tokens → {word_tokenize(test)}")

print("\n" + "=" * 60)
print("✅ Demo 01 complete — NLTK Tokenization")
print("=" * 60)
```

# DEMO 02 — Stopword Removal with NLTK and spaCy

```python
"""
Shows how to identify and filter stopwords using both libraries.
Demonstrates why stopword removal matters for NLP tasks.
"""

import nltk
nltk.download("punkt",     quiet=True)
nltk.download("punkt_tab", quiet=True)
nltk.download("stopwords", quiet=True)

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import spacy
```

```python
%pip install https://github.com/explosion/spacy-models/releases/download/en_core_web_sm-3.8.0/en_core_web_sm-3.8.0-py3-none-any.whl
```

```python
print("Loading spaCy model...")
nlp = spacy.load("en_core_web_sm")
print("✅ spaCy loaded\n")
```

```python
# ── SECTION 1: What are stopwords? ───────────────────────────────
print("=" * 60)
print("SECTION 1: What are English Stopwords?")
print("=" * 60)

nltk_sw = set(stopwords.words("english"))
spacy_sw = nlp.Defaults.stop_words

print(f"\n  NLTK  English stopwords : {len(nltk_sw)} words")
print(f"  spaCy English stopwords : {len(spacy_sw)} words")

# Show some examples
print(f"\n  Sample NLTK  stopwords : {sorted(list(nltk_sw))[:20]}")
print(f"  Sample spaCy stopwords : {sorted(list(spacy_sw))[:20]}")

# Words in spaCy but not NLTK
only_in_spacy = spacy_sw - {w.lower() for w in nltk_sw}
print(f"\n  Words in spaCy but NOT in NLTK ({len(only_in_spacy)} extra):")
print(f"  {sorted(list(only_in_spacy))[:25]}")
```

```python
# ── SECTION 2: Removing stopwords with NLTK ──────────────────────
print("\n" + "=" * 60)
print("SECTION 2: Stopword Removal with NLTK")
print("=" * 60)

text = (
    "Machine learning is a subset of artificial intelligence. "
    "It enables systems to automatically learn and improve from experience. "
    "The algorithms are trained on large datasets to make predictions."
)
print(f"\nOriginal text:\n  '{text}'\n")

# Tokenize
tokens = word_tokenize(text.lower())
print(f"After tokenization ({len(tokens)} tokens):")
print(f"  {tokens}\n")

# Filter stopwords + punctuation
nltk_filtered = [
    token for token in tokens
    if token.isalpha() and token not in nltk_sw
]
print(f"After NLTK stopword removal ({len(nltk_filtered)} tokens):")
print(f"  {nltk_filtered}")

removed_count = len([t for t in tokens if t.lower() in nltk_sw])
print(f"\n  → Removed {removed_count} stopwords ({removed_count/len(tokens)*100:.1f}% of tokens)")
```

```python
# ── SECTION 3: Removing stopwords with spaCy ─────────────────────
print("\n" + "=" * 60)
print("SECTION 3: Stopword Removal with spaCy")
print("=" * 60)

doc = nlp(text)

print(f"\nspaCy token analysis:")
print(f"  {'Token':<20} {'is_stop':<10} {'is_alpha':<10} {'pos_'}")
print(f"  {'-'*55}")
for token in doc:
    print(f"  {token.text:<20} {str(token.is_stop):<10} {str(token.is_alpha):<10} {token.pos_}")

spacy_filtered = [
    token.text.lower()
    for token in doc
    if not token.is_stop and token.is_alpha
]
print(f"\nAfter spaCy stopword removal ({len(spacy_filtered)} tokens):")
print(f"  {spacy_filtered}")
```

```python
# ── SECTION 4: NLTK vs spaCy comparison ─────────────────────────
print("\n" + "=" * 60)
print("SECTION 4: NLTK vs spaCy Stopword Comparison")
print("=" * 60)

test_sentence = "The running fox is quickly jumping over a lazy dog near the river."
test_sentence1 = "Home alone is a good movie"
print(f"\nTest sentence: '{test_sentence1}'\n")

# NLTK
nltk_tokens = word_tokenize(test_sentence1.lower())
nltk_result = [t for t in nltk_tokens if t.isalpha() and t not in nltk_sw]

# spaCy
spacy_doc = nlp(test_sentence1)
spacy_result = [t.text.lower() for t in spacy_doc if not t.is_stop and t.is_alpha]

print(f"  NLTK  result ({len(nltk_result)} tokens) : {nltk_result}")
print(f"  spaCy result ({len(spacy_result)} tokens) : {spacy_result}")
```

```python
# ── SECTION 5: Multilingual stopwords (NLTK) ─────────────────────
print("\n" + "=" * 60)
print("SECTION 5: Multilingual Stopwords (NLTK)")
print("=" * 60)

languages = ["english", "french", "german", "arabic", "spanish", "bengali"]
for lang in languages:
    try:
        sw = stopwords.words(lang)
        sample = sorted(sw)[:5]
        print(f"  {lang:<12}: {len(sw):>3} stopwords — e.g. {sample}")
    except Exception as e:
        print(f"  {lang:<12}: not available ({e})")
```

```python
# ── SECTION 6: When NOT to remove stopwords ───────────────────────
print("\n" + "=" * 60)
print("SECTION 6: When NOT to Remove Stopwords")
print("=" * 60)

print("""
  ❌ DON'T remove stopwords for:
     • Sentiment analysis: "not good" → removing "not" → "good" (WRONG!)
     • BERT/GPT models: these models need full context including stopwords
     • Machine translation: every word matters for grammar
     • QA / chatbots: "Is this correct?" — all words needed

  ✅ DO remove stopwords for:
     • TF-IDF / bag of words features
     • Topic modelling (LDA)
     • Document classification with classic ML
     • Text summarization pre-processing
     • Keyword extraction
""")

print("=" * 60)
print("✅ Demo 02 complete — Stopword Removal")
print("=" * 60)
```

# DEMO 03 — Stemming with NLTK

```python
"""
Demonstrates Porter, Snowball, and Lancaster stemmers.
Shows the difference between all three algorithms on
identical inputs, and compares stemming vs lemmatization.

NLTK downloads (auto-downloaded on first run):
    wordnet, averaged_perceptron_tagger_eng
"""

import nltk
nltk.download("wordnet",    quiet=True)
nltk.download("punkt",      quiet=True)
nltk.download("punkt_tab",  quiet=True)
nltk.download("stopwords",  quiet=True)
nltk.download("averaged_perceptron_tagger_eng", quiet=True)

from nltk.stem import PorterStemmer, SnowballStemmer, LancasterStemmer
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet, stopwords
from nltk.tokenize import word_tokenize
```

```python
porter    = PorterStemmer()
snowball  = SnowballStemmer("english")
lancaster = LancasterStemmer()
lemmatizer = WordNetLemmatizer()
sw = set(stopwords.words("english"))

print("✅ All stemmers loaded\n")


# ─── helpers ──────────────────────────────────────────────────────
def get_wordnet_pos(tag: str) -> str:
    m = {"J": wordnet.ADJ, "V": wordnet.VERB, "N": wordnet.NOUN, "R": wordnet.ADV}
    return m.get(tag[0], wordnet.NOUN)


def lemmatize_pos(word: str, pos_tag: str) -> str:
    return lemmatizer.lemmatize(word, get_wordnet_pos(pos_tag))
```

```python
# ═══════════════════════════════════════════════════════════════════
# SECTION 1: Basic stemmer comparison
# ═══════════════════════════════════════════════════════════════════
print("=" * 70)
print("SECTION 1: Porter vs Snowball vs Lancaster — Word by Word")
print("=" * 70)

test_words = [
    "running",   "studies",   "beautiful",  "generously",
    "connection","flies",     "wolves",     "went",
    "happily",   "maximum",   "caring",     "argued",
    "caresses",  "ponies",    "cats",       "trouble",
    "matting",   "mating",    "meetings",   "rats",
]

print(f"\n{'Word':<18} {'Porter':<18} {'Snowball':<18} {'Lancaster'}")
print(f"  {'-'*65}")
for w in test_words:
    p = porter.stem(w)
    s = snowball.stem(w)
    l = lancaster.stem(w)
    # flag words where algorithms disagree or produce non-words
    print(f"  {w:<18} {p:<18} {s:<18} {l}")
```

```python
# ═══════════════════════════════════════════════════════════════════
# SECTION 3: Multilingual Snowball
# ═══════════════════════════════════════════════════════════════════
print("=" * 70)
print("SECTION 3: Snowball Multilingual Support")
print("=" * 70)

print(f"\nSupported languages: {SnowballStemmer.languages}\n")

multilingual = [
    ("english",    ["running", "studies", "happiness"]),
    ("french",     ["manager",  "mangeons", "mangez"]),
    ("german",     ["laufen",  "läuft",   "gelaufen"]),
    ("spanish",    ["correr",  "corriendo","corremos"]),
    ("arabic",     ["درس",     "يدرس",    "دراسة"]),
]

for lang, words in multilingual:
    try:
        stemmer = SnowballStemmer(lang)
        stems   = [stemmer.stem(w) for w in words]
        print(f"  {lang:<12} {words} → {stems}")
    except Exception as e:
        print(f"  {lang:<12} → Error: {e}")

print("\n  ✅ Snowball supports 13+ languages — use it for multilingual NLP!")
```

```python
# ═══════════════════════════════════════════════════════════════════
# SECTION 4: Stemming on a full sentence
# ═══════════════════════════════════════════════════════════════════
print("\n" + "=" * 70)
print("SECTION 4: Full Sentence Stemming Pipeline")
print("=" * 70)

sentence = "The running scientists quickly studied better machine learning methods across multiple laboratories."
print(f"\nOriginal: '{sentence}'\n")

# Tokenize + lowercase
tokens = word_tokenize(sentence.lower())
# Remove stopwords + punctuation
clean_tokens = [t for t in tokens if t.isalpha() and t not in sw]

print(f"Tokens (no stopwords): {clean_tokens}\n")
print(f"{'Token':<18} {'Porter':<18} {'Snowball':<18} {'Lancaster'}")
print(f"  {'-'*65}")
for t in clean_tokens:
    p = porter.stem(t)
    s = snowball.stem(t)
    l = lancaster.stem(t)
    print(f"  {t:<18} {p:<18} {s:<18} {l}")

porter_result   = [porter.stem(t)   for t in clean_tokens]
snowball_result = [snowball.stem(t)  for t in clean_tokens]
lanc_result     = [lancaster.stem(t) for t in clean_tokens]

print(f"\nPorter   result: {porter_result}")
print(f"Snowball result: {snowball_result}")
print(f"Lancaster result: {lanc_result}")
```

```python
# Section 3 — spaCy POS-Aware Lemmatization

print("=" * 65)
print("SECTION 3: spaCy POS-Aware Lemmatization (Auto-detected)")
print("=" * 65)

print("\nspaCy automatically detects POS and uses it for lemmatization")
print("No manual POS mapping needed!\n")


texts = [
    "The wolves were running across the beautiful snowy fields.",
    "She is studying better strategies for the upcoming studies.",
    "The geese flew south while the children argued about going."
]


for text in texts:
    print(f"Input: '{text}'")
    print()

    print(f"{'Token':<15}{'POS':<10}{'Lemma'}")
    print("-" * 40)

    doc = nlp(text)

    clean_lemmas = []

    for token in doc:
        print(f"{token.text:<15}{token.pos_:<10}{token.lemma_}")

        if not token.is_stop and not token.is_punct:
            clean_lemmas.append(token.lemma_)

    print()
    print(f"Clean lemmas (no stopwords): {clean_lemmas}")
    print()
```

```python
# ── SECTION 4: Full pipeline comparison ──────────────────────────

def get_wordnet_pos(treebank_tag: str) -> str:
  """Convert Penn Treebank POS tag to WordNet POS constant."""
  if treebank_tag.startswith("J"):
      return wordnet.ADJ
  elif treebank_tag.startswith("V"):
      return wordnet.VERB
  elif treebank_tag.startswith("N"):
      return wordnet.NOUN
  elif treebank_tag.startswith("R"):
      return wordnet.ADV
  else:
      return wordnet.NOUN  # Default to noun


def lemmatize_with_pos_nltk(text: str) -> list[str]:
  """Full NLTK lemmatization pipeline with automatic POS detection."""
  tokens = word_tokenize(text.lower())
  pos_tags = nltk.pos_tag(tokens)
  return [
      lemmatizer.lemmatize(token, get_wordnet_pos(pos))
      for token, pos in pos_tags
      if token.isalpha()
  ]

print("\n" + "=" * 65)
print("SECTION 4: Complete Pipeline Comparison on Real Text")
print("=" * 65)

review = (
    "The running shoes are better than expected. "
    "My feet were hurting after walking, but these beautifully designed shoes "
    "helped significantly. The children also loved wearing them."
)
print(f"\nReview: '{review}'\n")

# NLTK pipeline
nltk_result = lemmatize_with_pos_nltk(review)
print(f"  NLTK lemmas  ({len(nltk_result)} tokens):")
print(f"  {nltk_result}")

# spaCy pipeline
spacy_doc = nlp(review)
spacy_result = [t.lemma_.lower() for t in spacy_doc if t.is_alpha and not t.is_stop]
print(f"\n  spaCy lemmas ({len(spacy_result)} tokens):")
print(f"  {spacy_result}")

print("\n" + "=" * 65)
print("✅ Demo 03 complete — Lemmatization")
print("=" * 65)
```

```python
import time
import nltk
from nltk import pos_tag
from nltk.corpus import wordnet

big_word_list = test_words * 500  # 10,000 words


def bench(fn):
    t0 = time.perf_counter()
    fn()
    return (time.perf_counter() - t0) * 1000


# ---------------- Stemmers ----------------
p_ms = bench(lambda: [porter.stem(w) for w in big_word_list])
s_ms = bench(lambda: [snowball.stem(w) for w in big_word_list])
l_ms = bench(lambda: [lancaster.stem(w) for w in big_word_list])


# ---------------- Lemmatization with POS ----------------
def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith("J"):
        return wordnet.ADJ
    elif treebank_tag.startswith("V"):
        return wordnet.VERB
    elif treebank_tag.startswith("N"):
        return wordnet.NOUN
    elif treebank_tag.startswith("R"):
        return wordnet.ADV
    return wordnet.NOUN


def pos_lemmatize():
    tags = pos_tag(big_word_list)
    return [
        lemmatizer.lemmatize(word, get_wordnet_pos(tag))
        for word, tag in tags
    ]


lem_ms = bench(pos_lemmatize)


print(f"""
10,000 words processed:

Porter      : {p_ms:.0f} ms ({p_ms/10:.2f} ms / 1000 words)
Snowball    : {s_ms:.0f} ms ({s_ms/10:.2f} ms / 1000 words)
Lancaster   : {l_ms:.0f} ms ({l_ms/10:.2f} ms / 1000 words)
Lemmatizer  : {lem_ms:.0f} ms ({lem_ms/10:.2f} ms / 1000 words)

Porter is {lem_ms/p_ms:.2f}× faster than POS-aware lemmatization.
Snowball is {lem_ms/s_ms:.2f}× faster than POS-aware lemmatization.
Lancaster is {lem_ms/l_ms:.2f}× faster than POS-aware lemmatization.
""")
```

When to use which — decision guide

```python
print("""
  USE STEMMING (Porter/Snowball) WHEN:
  ✅  Search engine / information retrieval (speed matters)
  ✅  TF-IDF features for classical ML (Naive Bayes, SVM, Logistic)
  ✅  Topic modelling (LDA) — vocabulary size reduction
  ✅  Large-scale document clustering (millions of docs)
  ✅  Multilingual NLP (Snowball supports 13+ languages)
  ✅  You need < 1ms per word processing speed

  USE LEMMATIZATION (NLTK/spaCy) WHEN:
  ✅  Sentiment analysis — 'better' → 'good' matters
  ✅  Chatbots and conversational AI
  ✅  Named entity recognition pipelines
  ✅  Any task where output must be a real dictionary word
  ✅  Fine-tuning BERT / transformer models

  USE NEITHER WHEN:
  ✅  Working with BERT, GPT, any HuggingFace transformer
       → These models use their own subword tokenization (BPE/WordPiece)
       → Pre-stemming/lemmatizing damages the token distribution
       → Feed raw (or minimally cleaned) text to transformer tokenizers
""")

print("=" * 70)
print("✅ Stemming and Lemmatization Demo complete!")
print("=" * 70)
```
